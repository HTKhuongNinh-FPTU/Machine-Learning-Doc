## Exercise 19: Data Augmentation: Random Crop
Apply random cropping to an image for data augmentation.

```python
# Your code here
import random

def random_crop(image, crop_height, crop_width):
    height, width, _ = image.shape

    x = random.randint(0, width - crop_width)
    y = random.randint(0, height - crop_height)

    cropped_image = image[y:y+crop_height, x:x+crop_width]
    return cropped_image

crop_height,crop_width = 100,100

cropped = random_crop(image_rgb,crop_height,crop_width)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.title('Original Image')

plt.subplot(1, 2, 2)
plt.imshow(cropped)
plt.title(f'Random Crop ({crop_width}x{crop_height})')

plt.show()
```

## Exercise 20: Data Augmentation: Color Jitter
Apply random color jitter (e.g., brightness, contrast) to an image.

```python
import random

# Exercise 20: Data Augmentation: Color Jitter
def color_jitter(image, contrast_limit, brightness_limit):
    # Fix: Added missing 'random' import and use correct parameters
    alpha = random.uniform(0.5, contrast_limit) # random contrast
    beta = random.randint(0, brightness_limit)  # random brightness

    jitter_image = cv2.convertScaleAbs(image, alpha=alpha, beta=beta)
    return jitter_image

contrast_lim, brightness_lim = 2, 50
jitter = color_jitter(image_rgb, contrast_lim, brightness_lim)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.title('Original Image')

plt.subplot(1, 2, 2)
plt.imshow(jitter)
plt.title('Color Jitter (Alpha/Beta random)')

plt.show()
```

## Exercise 22: Batch Process Images
Process a batch of images (e.g., resize and normalize) from a directory.

```python
import os
import cv2
import numpy as np

# Exercise 22: Fixed path to Colab environment
folder = '/content/'
size = (128, 128)

images_list = []

for file in os.listdir(folder):
    if file.endswith(("jpg", "png", "jpeg")):
        path = os.path.join(folder, file)
        img = cv2.imread(path)
        if img is not None:
            img = cv2.resize(img, size)
            img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
            img = img / 255.0
            images_list.append(img)

# Update the global images variable
images = np.array(images_list)
print("Total processed images:", len(images))
```

## Exercise 23: Prepare Data for ML
Convert images and labels into NumPy arrays for ML model input.

## Exercise 24: Save Processed Images
Save processed images to an output directory.

```python
# Exercise 24: Save Processed Images
import os

output_dir = 'output_images'
os.makedirs(output_dir, exist_ok=True)

# Save the first processed image (converting back to BGR for OpenCV saving)
if len(images) > 0:
    save_path = os.path.join(output_dir, 'processed_sample.jpg')
    # Scale back to 0-255 and convert to BGR
    img_to_save = (images[0] * 255).astype(np.uint8)
    img_to_save_bgr = cv2.cvtColor(img_to_save, cv2.COLOR_RGB2BGR)
    cv2.imwrite(save_path, img_to_save_bgr)
    print(f"Saved image to {save_path}")
```
